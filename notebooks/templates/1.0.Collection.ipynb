## Load and Check data

```python
import pandas as pd
import sqlite3

df_sample = pd.read_csv("../data/interim/extracted/sample.csv")
df_sub = pd.read_csv("../data/interim/extracted/substrate.csv")
df_cont = pd.read_csv("../data/interim/extracted/container.csv")
df_ferm = pd.read_csv("../data/interim/extracted/fermentation.csv")
df_measure = pd.read_csv("../data/interim/extracted/measurement.csv")

# Connect to SQLite and create a database
conn = sqlite3.connect("myco.db")
df_sample.to_sql("sample", conn, if_exists="replace", index=False)
df_sub.to_sql("substrates", conn, if_exists="replace", index=False)
df_cont.to_sql("containers", conn, if_exists="replace", index=False)
df_ferm.to_sql("fermentation", conn, if_exists="replace", index=False)
df_measure.to_sql("measurement", conn, if_exists="replace", index=False)

# 4. Commit changes and close connection
conn.commit()
conn.close()
```
